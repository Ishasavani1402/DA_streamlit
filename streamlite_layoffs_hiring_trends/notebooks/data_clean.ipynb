```python
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv(r'D:\Data Analytics\tech_layoffs_hiring_trends\datasets\tech_layoffs_hiring_trends.csv')
df.head(5)
```

```python
df = df.drop(columns=['stock_growth_percent', 'revenue_growth_percent', 'salary_budget_change'])
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ' , '_')
df.columns
```

```python
df.info()
print('=='*20)
print('null check')
print('=='*20)
print(df.isnull().sum())
print('=='*20)
print('duplicate check')
print('=='*20)
print(df.duplicated().sum())
print('=='*20)
print('data shape before cleaning')
print('=='*20)
print(df.shape)
```

```python
# check record_id is unique or not 
df['record_id'].nunique()
```

👆**here data shape = 12000 and record_id unique = 12000 means record id is all unique (primary key)**

```python
# remove whitespace from object columns
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.strip()
```

```python
# check unique values
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    vals = df[col].unique()
    print(f"\n{col} ({len(vals)} unique): {sorted(vals)}")
```

* to check negetive value exist in numeric column

```python
numeric_col = df.select_dtypes(include=['float64', 'int64'])

negative_cols = numeric_col.columns[(numeric_col < 0).any()]

print("Columns containing negative values:")
print(negative_cols.tolist())
```

* to check data range in numeric column

```python
bounds = {
    'layoff_percentage': (0, 100),
    'remote_jobs_percentage': (0, 100),
    'job_security_score': (1, 10),
    'ai_replacement_risk': (1, 10),
    'ai_adoption_level': (1, 10),
    'employee_sentiment': (1, 10),
}
for col, (lo, hi) in bounds.items():
    bad = df[(df[col] < lo) | (df[col] > hi)]
    print(f"{col}: {len(bad)} rows outside [{lo}, {hi}]")
```
