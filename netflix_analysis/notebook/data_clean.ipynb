```python
import pandas as pd
```

```python
df = pd.read_csv(r'D:\Data Analytics\DA_python_projects\netflix_analysis\dataset\Netflix Dataset.csv')
df.head(5)
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ','_')
df.columns
```

```python
df.info()
print("--"* 20)
print(f"null chheck : \n{df.isnull().sum()/ len(df)*100}")
print("--"* 20)
print(f"duplicate check : \n{df.duplicated().sum()}")
print("--"* 20)
print(f" data shape before clean: {df.shape}")
```

```python
df['show_id'].duplicated().any()
```

**here show id have duplicate that means show id is not primary key**

```python
# Display all rows involved in duplication
duplicate_rows = df[df.duplicated(keep=False)]
duplicate_rows
```

* drop duplicate 

```python
df = df.drop_duplicates().reset_index(drop=True)
print('after remove duplicate data shape')
df.shape
```

* handle missing value 

```python
df[['director', 'cast', 'country']] = df[['director', 'cast', 'country']].fillna('Unknown')

df[['rating']] = df[['rating']].fillna('Not Rated')

print('after handle null value')
print(df.isnull().sum())
```

```python
df = df.rename(columns={'release_date': 'date_added', 'type': 'genres'})
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce', format='mixed')
df = df.dropna(subset=['date_added'])
print('after handle null value')
print(df.isnull().sum())
```

```python
# remove white space from object 
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
```

```python
df = df.drop(columns=['description'])
```

```python
# check unique values
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    vals = df[col].unique()
    print(f"\n{col} ({len(vals)} unique): {sorted(vals)}")
```

```python
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
```

* new column that have only number from duration column

```python
df['duration_value'] = df['duration'].str.extract(r'(\d+)').astype(int)
```

* new column that have only season or minute from duration column

```python
df['duration_unit'] = df['duration'].str.extract(r'([A-Za-z]+)')
```

* create new column duration minute for category movie and column season for tv shows as we have two categpry tv shows and movies

```python
df['duration_min'] = df['duration_value'].where(df['category'] == 'Movie')
df['seasons'] = df['duration_value'].where(df['category'] == 'TV Show')
```

* for multiple country and genere we split it and keep only first occurece for better analysis

```python
df['primary_country'] = df['country'].str.split(',').str[0].str.strip()
df['primary_genre'] = df['genres'].str.split(',').str[0].str.strip()
```

```python
df.to_csv(r'D:\Data Analytics\DA_python_projects\netflix_analysis\dataset\clean_dataset.csv' , index=False)
```
